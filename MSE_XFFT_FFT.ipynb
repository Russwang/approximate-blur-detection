```python
# import all libraries
from tensorflow.keras.models import load_model
import matplotlib.pyplot as plt
import numpy as np
import cv2
import argparse
import os
from sklearn.metrics import mean_squared_error

block_size = [64, 128, 256] # 32, 64, 128, 256
reduce_size = 224
```

```python
# Define fft functions
def fft1d(x, n, model):
    if len(x) != n:
        raise ValueError(f'Input must be length {n}')
    input_vec = x.real[np.newaxis, :]
    pred = model.predict(input_vec, verbose=0)[0]
    result = pred[:n] + 1j * np.zeros_like(pred)
    return result
```

```python
block_size = 256
x256 = []
mse256 = []
j = 0
for reduce_size in range(16, 264, 16):
    
    model_path = os.path.join(os.getcwd(), 'model', f'1D_FFT_{block_size}_reduce_{reduce_size}.h5')
    model_1d = load_model(model_path)
    
    data = np.random.randn(block_size)
    
    ANN = fft1d(data, block_size, model_1d)
    FFT = np.fft.fft(data)

    diff = mean_squared_error(ANN.real, FFT.real)
    print(diff)
    mse256.append(diff)
    x256.append(reduce_size)
    j = j + 1
    
```
