```python
# import all libraries
from tensorflow.keras.models import load_model
import matplotlib.pyplot as plt
import numpy as np
import cv2
import argparse
import os
from sklearn.metrics import mean_squared_error
```

```python
def cplx_xfft(n, k):
    add = k*(n-1)+n*(k-1)
    mul = 2*k*n
    return add, mul

def cplx_ratio(n, k):
    r = (k*(n-1)+n*(k-1)+2*k*n)/(n*(n-1)+n*(n-1)+2*n*n)
```

```python
block_size = 128
x128 = []
mse128 = []
cplx128 = []
j = 0
for reduce_size in range(72, 128, 8):
        
    XANN_add, XANN_mul = cplx_xfft(block_size, reduce_size)
    ANN_add, ANN_mul = cplx_xfft(block_size, block_size)

    diff = ((ANN_add+ANN_mul)-(XANN_add+XANN_mul))/(ANN_add+ANN_mul)
    print(diff)
    mse128.append(diff*100)
    cplx128.append((XANN_add+XANN_mul)*160)
    x128.append(reduce_size)
    j = j + 1
```

```python
block_size = 256
x256 = []
mse256 = []
cplx256 = []
j = 0
for reduce_size in range(140, 240, 16):
        
    XANN_add, XANN_mul = cplx_xfft(block_size, reduce_size)
    ANN_add, ANN_mul = cplx_xfft(block_size, block_size)

    diff = ((ANN_add+ANN_mul)-(XANN_add+XANN_mul))/(ANN_add+ANN_mul)
    print(diff)
    mse256.append(diff*100)
    cplx256.append((XANN_add+XANN_mul)*40)
    x256.append(reduce_size)
    j = j + 1
```

```python
fig = plt.figure(figsize=(10,7))
plt.bar(x128, cplx128, width=6)
plt.bar(x256, cplx256, width=10)
plt.xlabel("K", fontsize="25")
plt.ylabel("Computational Complexity", fontsize="25")
plt.xticks(np.arange(72, 240, step=32), fontsize="25")
plt.yscale('log')
plt.legend(['N=128', 'N=256'], fontsize="25")
plt.grid('on')
plt.savefig('outimg/CPLX_XFFT.png', dpi=600)
plt.show()
```

```python
fig = plt.figure(figsize=(10,7))
plt.plot(x128, mse128, '-bo', linewidth=4.0, markersize=12)
plt.plot(x256, mse256, '-md', linewidth=4.0, markersize=12)
plt.xlabel("K", fontsize="25")
plt.ylabel("Complexity Reduction (%)", fontsize="25")
plt.xticks(np.arange(72, 240, step=32), fontsize="25")
plt.legend(['N=128', 'N=256'], fontsize="25")
plt.grid('on')
plt.savefig('outimg/CPLX_RATIO_XFFT.png', dpi=600)
plt.show()
```
